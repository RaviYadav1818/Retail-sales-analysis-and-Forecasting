# Notebook 8: Power BI Dashboard Preparation

Prepare datasets, star schema, KPIs and DAX measures.

```python
import pandas as pd

df=pd.read_csv('data/feature_engineered_retail_sales.csv',parse_dates=['Order_Date','Ship_Date'])
df.head()
```

```python
# Create Dimension Tables
dim_customer=df[['Customer_ID','Customer_Name','Segment']].drop_duplicates()
dim_product=df[['Product_ID','Product_Name','Category','Sub_Category']].drop_duplicates()
dim_location=df[['State','City','Region']].drop_duplicates()

date=pd.DataFrame({'Date':pd.date_range(df['Order_Date'].min(),df['Order_Date'].max())})
date['Year']=date['Date'].dt.year
date['Month']=date['Date'].dt.month
date['Month_Name']=date['Date'].dt.month_name()
date['Quarter']=date['Date'].dt.quarter

fact=df.copy()
```

```python
dim_customer.to_csv('data/dim_customer.csv',index=False)
dim_product.to_csv('data/dim_product.csv',index=False)
dim_location.to_csv('data/dim_location.csv',index=False)
date.to_csv('data/dim_date.csv',index=False)
fact.to_csv('data/fact_sales.csv',index=False)
print('Star schema CSV files exported.')
```

```python
dax_measures = '''
Total Sales = SUM(fact_sales[Sales])

Total Profit = SUM(fact_sales[Profit])

Total Orders = DISTINCTCOUNT(fact_sales[Order_ID])

Total Customers = DISTINCTCOUNT(fact_sales[Customer_ID])

Average Order Value = DIVIDE([Total Sales],[Total Orders])

Profit Margin % = DIVIDE([Total Profit],[Total Sales])*100

Total Quantity = SUM(fact_sales[Quantity])

YTD Sales = TOTALYTD([Total Sales],dim_date[Date])

MTD Sales = TOTALMTD([Total Sales],dim_date[Date])

Sales Growth % =
DIVIDE([Total Sales]-CALCULATE([Total Sales],DATEADD(dim_date[Date],-1,MONTH)),
CALCULATE([Total Sales],DATEADD(dim_date[Date],-1,MONTH)))
'''
print(dax_measures)
```
